```python
import os
import configparser
from dotenv import load_dotenv
```

```python
# @title パラメータ読み込み前の状態を取得
if "prev_os_environ" not in locals():
    prev_os_environ = set(os.environ)
```

```python
# @title Drive マウント(Colab環境用)
import sys

is_colab_env = "google.colab" in sys.modules

if is_colab_env:
    try:
        # TODO: [@TEMPLATE]はコンペ用に用意したフォルダ名に書き換える必要あり
        FOLDER_NAME = "@TEMPLATE"

        from google.colab import drive
        from google.colab import userdata
        os.environ["GOOGLE_USER_ADDRESS"] = userdata.get("GOOGLE_USER_ADDRESS")
        os.environ["GOOGLE_USER_PASSWORD"] = userdata.get("GOOGLE_USER_PASSWORD")
        drive.mount('/content/drive')

        %cd ./drive/MyDrive/kaggle/competition/{FOLDER_NAME}/working
    except:
        pass
```

```python
# @title パラメータ読み込み
load_dotenv()

config_ini = configparser.ConfigParser()
config_ini.read("../setting.ini", encoding="utf-8")

for key, value in config_ini.defaults().items():
    os.environ[key] = value
```
